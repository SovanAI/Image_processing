# Practice for to take a single image as a input and convert it to a tensor 

```python
from PIL import Image
from torchvision import transforms

# taking images as input 

sebrina = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/woman.jpeg")
sebrina.show()

print(sebrina.format, sebrina.size, sebrina.mode)
```

# Now resizing the image and trnasform the image to tensors

```python
from torchvision import transforms
# function to make the function to resize and apply operation to make the data ready for data augmentation

activation = transforms.Compose([
    transforms.Resize((256,256)),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor()
])

tensor_sebrina = activation(sebrina)
# love.show()
print(tensor_sebrina)
print(tensor_sebrina.shape)
print(tensor_sebrina.min())
print(tensor_sebrina.max())
```

# Normalization the tensor matrix data for CNN

```python
from torchvision import transforms
# basically we dont need to normalize any picture dataset at first
normalization = transforms.Normalize(
    mean = [0.5, 0.5 , 0.5],
    std = [0.5, 0.5, 0.5]
)

tensor_sebrina_normalize = normalization(tensor_sebrina)
print(tensor_sebrina_normalize)
```

```python
print(tensor_sebrina_normalize.min())
print(tensor_sebrina_normalize.max())
```
