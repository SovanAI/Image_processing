## Taking images as input

```python
from PIL import Image
image = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/cat.jpeg")
print(image.format, image.size, image.mode)
# image.show()
image2 = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/dog.jpeg")
print(image2.format, image2.size, image2.mode)
# image2.show()
```

## Resizing images to the proper size for classification matrix

```python
from torchvision import transforms
transform = transforms.Resize((224,224)) ## function to resize the image  H x W 

resize_image = transform(image)
resize_image2 = transform(image2)
print(resize_image.format, resize_image.size, resize_image.mode)
print(resize_image2.format, resize_image2.size, resize_image2.mode)
```

## Converting the images into tensors

```python
from torchvision import transforms
tensors_convertor = transforms.ToTensor()
## function to transform the resize image to tensors 

image_matrix = tensors_convertor(resize_image)
image2_matrix = tensors_convertor(resize_image2)

# printing the tensor matrixes for 
print(image_matrix)
print(image2_matrix)
```

# Center Crop , Random Crop and Resize function in PyTorch for model training data in neural networks

```python
from PIL import Image
from torchvision import transforms
cat_image = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/cat.jpeg")
dog_image = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/dog.jpeg")

# checking for the dataset "cat image"
# cat_image.show() 

# checking for the dataset "dog.jpeg"
# dog_image.show()

# function to resize the image pixels and convert it to tensors

transform_tensors_resize = transforms.Compose(
    [
        transforms.Resize((224,224)),
        transforms.ToTensor()
    ]
)

# function to crop the image pixels and convert it to tensors 

transform_tensors_crop = transforms.Compose(
    [
        transforms.CenterCrop((224,224)),
        transforms.ToTensor() 
    ]
)

# result using the Resize function 

cat_image_resize_matrix = transform_tensors_resize(cat_image)
dog_image_resize_matrix = transform_tensors_resize(dog_image)

# result using the Crop function

cat_image_crop_matrix = transform_tensors_crop(cat_image)
dog_image_crop_matrix = transform_tensors_crop(dog_image)

# printing the matrixes using differt functions

# printing the resize matrix

print(cat_image_resize_matrix )
print(dog_image_resize_matrix)

# printing the reshape matrix

print(cat_image_crop_matrix)
print(dog_image_crop_matrix)
```

## Using the Normalization function and other distributions for the image classification of the Data

```python
from PIL import Image
from torchvision import transforms , datasets 

# The dataset librery using for a large datasets and tensor labels 

dog_image = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/dog.jpeg")
cat_image = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/cat.jpeg")

# The image datset show 

# dog_image.show() 
# cat_image.show()

# datasets transform to teensors

tensor = transforms.Compose([

    transforms.RandomCrop(220),
    transforms.ToTensor()

])

# Normalization of tensors

normalization = transforms.Normalize(
    mean = [0.5,0.5,0.5],
    std = [0.5,0.5,0.5]
)

# using the functins normalize the tensors

normalize_tensor_cat = normalization(tensor(cat_image)) 
normalize_tensor_dog = normalization(tensor(dog_image)) 

# printing the values of tensors after the normalization

print(normalize_tensor_cat)
print(normalize_tensor_dog)

# printing the shapes of the tensor matrixes

print(normalize_tensor_cat.shape)
print(normalize_tensor_dog.shape)
```

# All needed tools to label and mark datasets to train models and data augmentations

```python
from torchvision import transforms
from PIL import Image

test_data = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/dog.jpeg")
data_label_module = transforms.Compose(
    [
        transforms.RandomRotation(10), # Rotates an image or tensor at random location at any directional degrees
        transforms.Resize(220),   # Resize the pixes of the image 
        transforms.RandomCrop(220), # this function crops a tensor at random location
        transforms.RandomHorizontalFlip(), # lips an input image horizontally (left-to-right reflection) based on a specified probability
        transforms.ToTensor()  # convert an image to tensor
    ]
)

# Getting differnt kind of augmented data 

# specially for the dog image
augmented_data = data_label_module(test_data)
print(augmented_data)

```

# Augmented Data for the cat dataset

```python
cat_image = Image.open("/home/sovan-rajbanshi/Projects/Image-processing/datasets/cat.jpeg")
augmented_data_cat = data_label_module(cat_image)
print(augmented_data_cat)
```
